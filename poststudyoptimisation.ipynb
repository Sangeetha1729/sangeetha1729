```python
import numpy as np
import sympy as sp
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Define symbols
t = sp.symbols('t')
alpha = sp.Function('alpha')(t)
theta = sp.Function('theta')(t)
phi = sp.Function('phi')(t)

# Constants
II = 1
M = 1
m = 0.5
a = 1
b = 0.8
c = 0.1
A = 1
omega = 0.1
mu = 10
Lambda = 0.0001
g = 1
theta_max = A / 2

# Define functions
x = -a * sp.sin(alpha) + b * sp.sin(alpha + theta) + c * sp.sin(alpha + theta + phi)
y = a * sp.cos(alpha) - b * sp.cos(alpha + theta) - c * sp.cos(alpha + theta + phi)
zeta = -a * sp.sin(alpha) + b * sp.sin(alpha + theta)
eta = a * sp.cos(alpha) - b * sp.cos(alpha + theta)

# Potential energy function
U_theta = 0.000001 * (1 / ((theta - theta_max)**6) + 1 / ((theta + theta_max)**6))

# Calculate kinetic energies and inertia
KEm = (m / 2) * (sp.diff(x, t)**2 + sp.diff(y, t)**2)
KEM = (M / 2) * (sp.diff(zeta, t)**2 + sp.diff(eta, t)**2)
inertia = (II / 2) * sp.diff(alpha, t)**2

# Potential energy
PEm = m * g * y
PEM = M * g * eta

# Wave restoring force
waveRestoring = 10 * (sp.sin(omega * t) - alpha)**2

# Lagrangian
L = sp.simplify(KEm + KEM + inertia - (PEm + PEM + U_theta + waveRestoring))

# Velocity for Rayleigh dissipation function
v = sp.Abs(sp.diff(sp.sqrt(U_theta), theta)).subs(theta, theta)

# Rayleigh dissipation function
R = (1 / 2) * mu * (sp.diff(alpha, t) - A * sp.sin(omega * t))**2 + (1 / 2) * v * sp.diff(theta, t)**2 + (1 / 2) * Lambda * sp.diff(phi, t)**2

# Equations of motion
eqn1 = sp.simplify(sp.diff(sp.diff(L, sp.diff(alpha, t)), t) - sp.diff(L, alpha) + sp.diff(R, sp.diff(alpha, t)))
eqn2 = sp.simplify(sp.diff(sp.diff(L, sp.diff(theta, t)), t) - sp.diff(L, theta) + sp.diff(R, sp.diff(theta, t)))
eqn3 = sp.simplify(sp.diff(sp.diff(L, sp.diff(phi, t)), t) - sp.diff(L, phi) + sp.diff(R, sp.diff(phi, t)))

# Convert sympy expressions to lambda functions for numerical solving
f_eqn1 = sp.lambdify((t, alpha, theta, phi, sp.diff(alpha, t), sp.diff(theta, t), sp.diff(phi, t)), eqn1)
f_eqn2 = sp.lambdify((t, alpha, theta, phi, sp.diff(alpha, t), sp.diff(theta, t), sp.diff(phi, t)), eqn2)
f_eqn3 = sp.lambdify((t, alpha, theta, phi, sp.diff(alpha, t), sp.diff(theta, t), sp.diff(phi, t)), eqn3)

# Define functions for the right-hand side of the ODEs
def equations(t, y):
    alpha, theta, phi, alpha_dot, theta_dot, phi_dot = y
    alpha_ddot = f_eqn1(t, alpha, theta, phi, alpha_dot, theta_dot, phi_dot)
    theta_ddot = f_eqn2(t, alpha, theta, phi, alpha_dot, theta_dot, phi_dot)
    phi_ddot = f_eqn3(t, alpha, theta, phi, alpha_dot, theta_dot, phi_dot)
    return [alpha_dot, theta_dot, phi_dot, alpha_ddot, theta_ddot, phi_ddot]

# Initial conditions
y0 = [0, 0, 0, A * omega, -A * omega, 0]

# Time span
t_span = (0, 100)

# Solving the ODEs
sol = solve_ivp(equations, t_span, y0, method='RK45', t_eval=np.linspace(0, 100, 1000))

# Extracting results
alpha_sol = sol.y[0]
theta_sol = sol.y[1]
phi_sol = sol.y[2]

# Plot alpha
plt.plot(sol.t, alpha_sol, label='alpha(t)')
plt.plot(sol.t, A * np.sin(omega * sol.t), label='beta(t)', color='red')
plt.legend()
plt.show()

# Plot theta
plt.plot(sol.t, theta_sol)
plt.show()

# Plot phi
plt.plot(sol.t, phi_sol)
plt.show()

# Parametric plots
plt.plot(alpha_sol, theta_sol)
plt.show()

plt.plot(alpha_sol, phi_sol)
plt.show()
```
